```python
engine_psql = create_engine("postgresql+psycopg2://postgres:12345678@localhost:5432/walmart_db")

try:
    engine_psql.connect()
    print("Connection Successful to PSQL")
except Exception as e:
    print("Unable to connect:", e)
```

```python
df.to_sql(name='walmart', con=engine_psql, if_exists='replace', index=False)
```
